# Meteor Scatter 3D Explorer

VHF forward meteor scatter: where a meteor trail has to be, and how it has to be moving, to produce a given Doppler shift on a TV carrier received off-air.

**How to use this notebook**
1. Choose **Run All** (toolbar at the top). It runs the cells below. The form appears under Step 2, filled in with the WDPN (Philadelphia) to Washington DC example, and Step 3 then draws its results.
2. Edit the fields in the form and click **Update and run**. The results appear in the four tabs under the buttons: *Summary*, *Static plots*, *Tables* and *3D explorer*. Nothing else needs to be run again.

Doppler model: f_D = -(v/lambda) * vhat . (uT + uR), where uT and uR point from the transmitter and receiver to the reflection point. Positive = path shortening.
